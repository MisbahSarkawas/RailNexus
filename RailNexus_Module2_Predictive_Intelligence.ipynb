```python

import pandas as pd
import os

info = pd.read_csv("/content/train_info.csv")
schedule = pd.read_csv(
    "/content/railway_data/train_schedule.csv"
)

# Standardize column-name whitespace
info.columns = info.columns.str.strip()
schedule.columns = schedule.columns.str.strip()

print("RAILNEXUS - DATA VALIDATION")
print("-" * 40)

# 1. Check train-number matching
info_trains = set(info["Train_No"].dropna())
schedule_trains = set(schedule["Train_No"].dropna())

print("Train numbers only in train_info:",
      len(info_trains - schedule_trains))

print("Train numbers only in schedule:",
      len(schedule_trains - info_trains))

print("Train numbers shared by both:",
      len(info_trains & schedule_trains))

# 2. Check schedule time formats
for col in ["Arrival_time", "Departure_Time"]:
    if col in schedule.columns:
        parsed = pd.to_datetime(
            schedule[col].astype(str),
            format="%H:%M:%S",
            errors="coerce"
        )

        print(f"\n{col}")
        print("Valid time values:", parsed.notna().sum())
        print("Invalid time values:", parsed.isna().sum())
        print("Examples:", schedule[col].head(5).tolist())

# 3. Check whether a train has multiple route records
if "Route_Number" in schedule.columns:
    routes = schedule.groupby("Train_No")[
        "Route_Number"
    ].nunique()

    print("\nTrains with multiple route numbers:",
          (routes > 1).sum())

print("\nValidation completed.")
```

```python

import pandas as pd
import os

input_path = "/content/railway_data/train_schedule.csv"
output_dir = "/content/railway_data/processed"
os.makedirs(output_dir, exist_ok=True)

schedule = pd.read_csv(input_path)
schedule.columns = schedule.columns.str.strip()

# Standardize text fields without changing original files
for col in ["Station_Code", "Station_Name"]:
    if col in schedule.columns:
        schedule[col] = schedule[col].astype("string").str.strip()

# Convert scheduled times into time objects
for col in ["Arrival_time", "Departure_Time"]:
    if col in schedule.columns:
        schedule[col] = pd.to_datetime(
            schedule[col].astype(str),
            format="%H:%M:%S",
            errors="coerce"
        ).dt.time

# Check the distance column
if "Distance" in schedule.columns:
    schedule["Distance"] = pd.to_numeric(
        schedule["Distance"], errors="coerce"
    )
    print("Missing distances:", schedule["Distance"].isna().sum())
    print("Negative distances:", (schedule["Distance"] < 0).sum())

# Save a processed copy
output_path = os.path.join(
    output_dir, "train_schedule_cleaned.csv"
)
schedule.to_csv(output_path, index=False)

print("\nProcessed dataset saved to:", output_path)
print("Rows and columns:", schedule.shape)
print("Duplicate rows:", schedule.duplicated().sum())
print("\nProcessed sample:")
display(schedule.head())
```

```python

import pandas as pd

path = "/content/railway_data/processed/train_schedule_cleaned.csv"
df = pd.read_csv(path)

df = df.sort_values(["Train_No", "SN"]).copy()

for col in ["Arrival_time", "Departure_Time"]:
    df[f"{col}_is_midnight"] = (
        df[col].astype(str).str.strip() == "00:00:00"
    )

# Identify first and last recorded stops per train
first_rows = df.groupby("Train_No").head(1)
last_rows = df.groupby("Train_No").tail(1)

print("MIDNIGHT TIME INVESTIGATION")
print("-" * 40)

for col in ["Arrival_time", "Departure_Time"]:
    flag = f"{col}_is_midnight"

    print(f"\n{col}")
    print("Total midnight values:", df[flag].sum())
    print("Midnight values at first stop:",
          first_rows[flag].sum())
    print("Midnight values at last stop:",
          last_rows[flag].sum())

    print("\nExamples of midnight records:")
    display(
        df.loc[
            df[flag],
            ["Train_No", "SN", "Station_Name", col]
        ].head(10)
    )

print("\nInvestigation completed.")
```

```python

import pandas as pd
import os

input_path = "/content/railway_data/processed/train_schedule_cleaned.csv"
output_dir = "/content/railway_data/processed"

df = pd.read_csv(input_path)
df = df.sort_values(["Train_No", "SN"]).copy()

# Numeric distance and route sequence
df["Distance"] = pd.to_numeric(df["Distance"], errors="coerce")
df["SN"] = pd.to_numeric(df["SN"], errors="coerce")

# Flag midnight values without changing the original times
df["arrival_midnight_flag"] = (
    df["Arrival_time"].astype(str) == "00:00:00"
).astype(int)

df["departure_midnight_flag"] = (
    df["Departure_Time"].astype(str) == "00:00:00"
).astype(int)

# Record the previous stop's distance for each train
df["previous_distance_km"] = (
    df.groupby("Train_No")["Distance"].shift(1)
)

df["distance_since_previous_stop_km"] = (
    df["Distance"] - df["previous_distance_km"]
)

# Identify the first and last scheduled stops
df["is_first_stop"] = (
    df.groupby("Train_No")["SN"].transform("min") == df["SN"]
).astype(int)

df["is_last_stop"] = (
    df.groupby("Train_No")["SN"].transform("max") == df["SN"]
).astype(int)

output_path = os.path.join(
    output_dir, "timetable_features.csv"
)
df.to_csv(output_path, index=False)

print("Timetable features saved:", output_path)
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nNew feature summary:")
display(df[[
    "Train_No", "SN", "Station_Code", "Distance",
    "distance_since_previous_stop_km",
    "arrival_midnight_flag", "departure_midnight_flag",
    "is_first_stop", "is_last_stop"
]].head(10))

print("\nMissing previous distances:",
      df["previous_distance_km"].isna().sum())
```

```python
import pandas as pd

df.columns = df.columns.str.strip()

df["Scheduled_Arrival"] = pd.to_datetime(
    df["Scheduled_Arrival"], format="%H:%M"
)
df["Actual_Arrival"] = pd.to_datetime(
    df["Actual_Arrival"], format="%H:%M"
)

df["Calculated_Delay_Minutes"] = (
    df["Actual_Arrival"] - df["Scheduled_Arrival"]
).dt.total_seconds() / 60

print("Dataset rows:", len(df))
print("\nDelay comparison:")
display(df[[
    "Train_Name",
    "Scheduled_Arrival",
    "Actual_Arrival",
    "Delay_Minutes",
    "Calculated_Delay_Minutes"
]])

print(
    "\nDo recorded delays match calculated delays?",
    (df["Delay_Minutes"] == df["Calculated_Delay_Minutes"]).all()
)
```
